```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

class linearregression: #定义线性回归
    def __init__(self): 
        self.theta = None #初始化权重

    def fit(self, x_b, y): 
        m_features = x_b.shape[1] #单位矩阵
        alpha = 1.0  #正则强度
        self.theta = np.linalg.inv(x_b.T @ x_b + alpha * np.eye(m_features)) @ x_b.T @ y #多元线性回归矩阵方程

    def predict(self, x_b):
        return x_b @ self.theta #用训练结果预测

#加载数据集
url = "https://raw.githubusercontent.com/ageron/handson-ml2/master/datasets/housing/housing.csv"
data = pd.read_csv(url).dropna()


feature = data.drop("median_house_value", axis=1).select_dtypes(include=[np.number]).values #房子特征集合
value = data["median_house_value"].values / 100000 #房价，单位万元

feature_train, feature_temp, value_train, value_temp = train_test_split(feature, value, test_size=0.4, random_state=0) #分为训练集和临时集，比例6：4
feature_valid, feature_test, value_valid, value_test = train_test_split(feature_temp, value_temp,test_size=0.5, random_state=0) #设置验证集和测试集，比例1：1；最终训练集：验证集：测试集=6：2：2

scaler = StandardScaler() #标准化工具
feature_train_scaled = scaler.fit_transform(feature_train) #训练集学习均值和方差并使用学习到的规则做标准化
feature_valid_scaled = scaler.transform(feature_valid) #验证集使用训练集学习到的规则做标准化
feature_test_scaled = scaler.transform(feature_test) #测试集使用训练集学习到的规则做标准化

feature_train_b = np.c_[np.ones((feature_train_scaled.shape[0], 1)), feature_train_scaled] #添加截距
feature_valid_b = np.c_[np.ones((feature_valid_scaled.shape[0], 1)), feature_valid_scaled] #同上
feature_test_b = np.c_[np.ones((feature_test_scaled.shape[0], 1)), feature_test_scaled] #同上

model =linearregression() #定义模型
model.fit(feature_train_b, value_train) #训练模型

#预测结果
value_train_pred = model.predict(feature_train_b)
value_valid_pred = model.predict(feature_valid_b)
value_test_pred = model.predict(feature_test_b)

#计算三个集真实值和预测值的MSE和R²
mse_train = mean_squared_error(value_train, value_train_pred)
r2_train = r2_score(value_train,value_train_pred)

mse_valid = mean_squared_error(value_valid, value_valid_pred)
r2_valid = r2_score(value_valid, value_valid_pred)

mse_test = mean_squared_error(value_test, value_test_pred)
r2_test = r2_score(value_test, value_test_pred)

#输出结果
print("train MSE: ", mse_train, ", R²: ", r2_train)
print("valid MSE: ", mse_valid, ", R²: ", r2_valid)
print("test MSE: ", mse_test, ", R²: ", r2_test)
print("trained weights (theta):", model.theta)

#真实值与预测值可视化
plt.figure(figsize=(18, 5))

#训练集可视化
plt.subplot(1, 3, 1)
plt.scatter(value_train, value_train_pred, alpha=0.2, color='blue')
plt.plot([value.min(), value.max()], [value.min(), value.max()], 'r--')
plt.xlabel("True Value")
plt.ylabel("Pred Value")
plt.title("Train Set")
plt.grid(True)

#验证集可视化
plt.subplot(1, 3, 2)
plt.scatter(value_valid, value_valid_pred, alpha=0.2, color='blue')
plt.plot([value.min(), value.max()], [value.min(), value.max()], 'r--')
plt.xlabel("True Value")
plt.ylabel("Pred Value")
plt.title("Valid Set")
plt.grid(True)

#测试集可视化
plt.subplot(1, 3, 3)
plt.scatter(value_test, value_test_pred, alpha=0.2, color='blue')
plt.plot([value.min(), value.max()], [value.min(), value.max()], 'r--')
plt.xlabel("True Value")
plt.ylabel("Pred Value")
plt.title("Test Set")
plt.grid(True)

plt.tight_layout()
plt.show()
```
